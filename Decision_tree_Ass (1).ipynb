# **----Theoretical Question----**

1. What is a Decision Tree, and how does it work?
- Ans:- A Decision Tree is a supervised learning algorithm that uses a tree-like model to classify data or make predictions. It works by recursively partitioning the data into smaller subsets based on the features of the data. Each internal node represents a feature or attribute, and each leaf node represents a class label or prediction.



















2. What are impurity measures in Decision Trees?
- Ans :- Impurity measures are used to evaluate the quality of a split in a Decision Tree. Common impurity measures include Gini Impurity, Entropy, and Variance. These measures help determine the best feature to split on at each node.

3. What is the mathematical formula for Gini Impurity?
- Ans:- The Gini Impurity formula is:

      Gini Impurity = 1 - ∑(p_i^2)

      where p_i is the proportion of instances in the node that belong to class i. This formula calculates the probability of incorrectly classifying a randomly chosen instance from the node.

4. What is the mathematical formula for Entropy?
- Ans:- The Entropy formula is:

      Entropy = -∑(p_i * log2(p_i))

     where p_i is the proportion of instances in the node that belong to class i. This formula calculates the amount of uncertainty or randomness in the node.

5. What is Information Gain, and how is it used in Decision Trees?
- Ans:- Information Gain is the reduction in impurity or uncertainty after splitting a node. It is calculated as the difference between the impurity of the parent node and the weighted average of the impurities of the child nodes. The feature with the highest Information Gain is chosen as the splitting feature.



6. What is the difference between Gini Impurity and Entropy?
-  Ans:- Both Gini Impurity and Entropy are impurity measures used in Decision Trees. Gini Impurity is more sensitive to the proportion of instances in each class, while Entropy is more sensitive to the uncertainty or randomness of the data. In practice, both measures often produce similar results.

7. What is the mathematical explanation behind Decision Trees?
- Ans:- Decision Trees are based on the concept of recursive partitioning, where the data is split into smaller subsets based on the features. The goal is to minimize the impurity or uncertainty of the data at each node. This is achieved by selecting the feature that results in the largest reduction in impurity or uncertainty.


8. What is Pre-Pruning in Decision Trees?
- Ans :- Pre-Pruning is a technique used to prevent overfitting in Decision Trees by stopping the growth of the tree before it becomes too complex. This is done by setting a maximum depth or a minimum number of instances required to split a node.

9. What is Post-Pruning in Decision Trees?
-   Ans:- Post-Pruning is a technique used to prevent overfitting in Decision Trees by removing branches or nodes that do not contribute significantly to the model's performance. This is done after the tree has been fully grown.


10. What is the difference between Pre-Pruning and Post-Pruning?
-   Ans:- Pre-Pruning prevents overfitting by stopping the growth of the tree, while Post-Pruning prevents overfitting by removing branches or nodes after the tree has been fully grown. Pre-Pruning is often faster and more efficient, while Post-Pruning can result in a more optimal tree.



11. What is a Decision Tree Regressor?
-  Ans:- A Decision Tree Regressor is a type of Decision Tree used for regression tasks, where the target variable is continuous. It works by predicting the average value of the target variable for each leaf node.


12.
What are the advantages and disadvantages of Decision Trees?
-  Ans:- Advantages:

  - Easy to interpret and visualize
  - Can handle categorical features
  - Can handle missing values
  - Fast training time

- Disadvantages:

  - Can suffer from overfitting
  - Can be sensitive to noise in the data
  - May not perform well on complex datasets



13. How does a Decision Tree handle missing values?
-  Ans:- Decision Trees can handle missing values by either ignoring them or treating them as a separate category. Some Decision Tree algorithms, such as CART, use a surrogate split method to handle missing values.


14.
How does a Decision Tree handle categorical features?
-  Ans:- Decision Trees can handle categorical features by treating them as nominal or ordinal variables. For nominal variables, the Decision Tree algorithm will consider all possible splits. For ordinal variables, the Decision Tree algorithm will consider splits that preserve the order of the categories.


15. What are some real-world applications of Decision Trees?
-  Ans:- Decision Trees have many real-world applications, including:

  - Credit risk assessment
  - Medical diagnosis
  - Customer segmentation
  - Predicting customer churn
  - Image classification
  - Financial forecasting

```python
# 1. Train Decision Tree Classifier on Iris dataset and print accuracy
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.3, random_state=42)

clf = DecisionTreeClassifier()
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
print("Model accuracy (default criterion):", accuracy_score(y_test, y_pred))
```

```python
# 2. Decision Tree with Gini Impurity and print feature importances
clf_gini = DecisionTreeClassifier(criterion='gini')
clf_gini.fit(X_train, y_train)
print("Feature importances (Gini):", clf_gini.feature_importances_)
```

```python
# 3. Decision Tree with Entropy and print accuracy
clf_entropy = DecisionTreeClassifier(criterion='entropy')
clf_entropy.fit(X_train, y_train)
y_pred_entropy = clf_entropy.predict(X_test)
print("Model accuracy (Entropy):", accuracy_score(y_test, y_pred_entropy))
```

```python
# 4. Decision Tree Regressor on housing dataset
from sklearn.datasets import fetch_california_housing
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error

housing = fetch_california_housing()
X_train, X_test, y_train, y_test = train_test_split(housing.data, housing.target, test_size=0.3, random_state=42)

regressor = DecisionTreeRegressor()
regressor.fit(X_train, y_train)
y_pred_reg = regressor.predict(X_test)
print("MSE (Decision Tree Regressor):", mean_squared_error(y_test, y_pred_reg))
```

```python
#6
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.3, random_state=42)

# Fully grown tree
clf_full = DecisionTreeClassifier()
clf_full.fit(X_train, y_train)
acc_full = accuracy_score(y_test, clf_full.predict(X_test))

# Tree with max_depth=3
clf_depth3 = DecisionTreeClassifier(max_depth=3)
clf_depth3.fit(X_train, y_train)
acc_depth3 = accuracy_score(y_test, clf_depth3.predict(X_test))

print(f"Accuracy (full tree): {acc_full:.4f}")
print(f"Accuracy (max_depth=3): {acc_depth3:.4f}")
```

```python
#7
# Default tree
clf_default = DecisionTreeClassifier()
clf_default.fit(X_train, y_train)
acc_default = accuracy_score(y_test, clf_default.predict(X_test))

# Tree with min_samples_split=5
clf_split5 = DecisionTreeClassifier(min_samples_split=5)
clf_split5.fit(X_train, y_train)
acc_split5 = accuracy_score(y_test, clf_split5.predict(X_test))

print(f"Accuracy (default): {acc_default:.4f}")
print(f"Accuracy (min_samples_split=5): {acc_split5:.4f}")

```

```python
from sklearn.datasets import fetch_california_housing, load_iris
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_squared_error, accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
# 11. Train Decision Tree Regressor with max_depth=5 and compare with full tree
housing = fetch_california_housing()
X_train, X_test, y_train, y_test = train_test_split(housing.data, housing.target, test_size=0.3, random_state=42)

reg_full = DecisionTreeRegressor()
reg_full.fit(X_train, y_train)
y_pred_full = reg_full.predict(X_test)
mse_full = mean_squared_error(y_test, y_pred_full)

reg_limited = DecisionTreeRegressor(max_depth=5)
reg_limited.fit(X_train, y_train)
y_pred_limited = reg_limited.predict(X_test)
mse_limited = mean_squared_error(y_test, y_pred_limited)

print("MSE (full tree):", mse_full)
print("MSE (max_depth=5):", mse_limited)
```

```python
# 12. Cost Complexity Pruning and its effect on accuracy
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.3, random_state=42)

clf = DecisionTreeClassifier(random_state=42)
path = clf.cost_complexity_pruning_path(X_train, y_train)
ccp_alphas = path.ccp_alphas
clfs = [DecisionTreeClassifier(random_state=42, ccp_alpha=alpha).fit(X_train, y_train) for alpha in ccp_alphas]

test_scores = [accuracy_score(y_test, clf.predict(X_test)) for clf in clfs]

plt.figure(figsize=(8, 4))
plt.plot(ccp_alphas, test_scores, marker='o')
plt.xlabel('ccp_alpha')
plt.ylabel('Accuracy')
plt.title('Effect of CCP on Accuracy')
plt.grid(True)
plt.show()
```

```python
# 13. Evaluate Precision, Recall, F1-Score
clf = DecisionTreeClassifier()
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)

print("Precision:", precision_score(y_test, y_pred, average='macro'))
print("Recall:", recall_score(y_test, y_pred, average='macro'))
print("F1 Score:", f1_score(y_test, y_pred, average='macro'))
```

```python
# 14. Confusion Matrix Visualization
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=iris.target_names, yticklabels=iris.target_names)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```

```python
# 15. GridSearchCV for optimal max_depth and min_samples_split
param_grid = {
    'max_depth': [2, 3, 4, 5, None],
    'min_samples_split': [2, 5, 10]
}
grid_search = GridSearchCV(DecisionTreeClassifier(), param_grid, cv=5)
grid_search.fit(X_train, y_train)

print("Best parameters:", grid_search.best_params_)
print("Best cross-validation score:", grid_search.best_score_)

best_clf = grid_search.best_estimator_
y_pred_best = best_clf.predict(X_test)
print("Test accuracy with best parameters:", accuracy_score(y_test, y_pred_best))
```
